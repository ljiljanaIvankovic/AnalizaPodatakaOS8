```python
# =====================================================================
# DEO 1 & 2: PODACI I AI LOGIKA (K-Najbližih suseda)
# =====================================================================

# 1. Naš skup podataka (Boja: 1=žuta, 2=zelena, 3=crvena | Veličina: 1=mala, 2=srednja, 3=velika)
primeri = [
    [3, 2],  # crvena, srednja  --> 'jabuka'
    [3, 3],  # crvena, velika   --> 'jabuka'
    [2, 2],  # zelena, srednja  --> 'jabuka'
    [2, 3],  # zelena, velika   --> 'jabuka'
    [1, 1],  # zuta, mala       --> 'banana'
    [1, 2],  # zuta, srednja    --> 'banana'
    [1, 3],  # zuta, velika     --> 'banana'
    [2, 1],  # zelena, mala     --> 'banana'
]

# Tačni odgovori (0 = jabuka, 1 = banana)
odgovori = [0, 0, 0, 0, 1, 1, 1, 1]

print("Podaci uspešno učitani!")
print(f"Ukupan broj primera u bazi: {len(primeri)}\n")


# 2. Ručna pravila (Kako bismo mi programirali bez AI-ja)
def pogodi_rucno(boja, velicina):
    if boja == 1:  # žuta
        return "banana"
    elif boja == 3:  # crvena
        return "jabuka"
    elif boja == 2 and velicina == 1:  # zelena i mala
        return "banana"
    else:
        return "jabuka"


# 3. AI funkcije (Računanje matematičke sličnosti)
def rastojanje(primer1, primer2):
    """Računa koliko su dve voćke slične. Manji broj = veća sličnost."""
    razlika_boje = (primer1[0] - primer2[0]) ** 2
    razlika_vel = (primer1[1] - primer2[1]) ** 2
    return razlika_boje + razlika_vel


def ai_pogodi(novo_voce, primeri, odgovori):
    """AI pretražuje pamćenje i traži najsličniju voćku."""
    najmanje_rastojanje = 999  # Počinjemo sa velikim brojem
    najslicniji_odgovor = None

    for primer, odgovor in zip(primeri, odgovori):
        r = rastojanje(novo_voce, primer)
        if r < najmanje_rastojanje:
            najmanje_rastojanje = r
            najslicniji_odgovor = odgovor

    naziv = "jabuka" if najslicniji_odgovor == 0 else "banana"
    return naziv, najmanje_rastojanje


# === TESTIRANJE AI-ja NA POZNATIM PRIMERIMA ===
print("--- Pokrećemo AI testiranje na poznatim voćkama ---")
tacno = 0
for i, (primer, pravi) in enumerate(zip(primeri, odgovori)):
    predvidjanje, _ = ai_pogodi(primer, primeri, odgovori)
    pravi_naziv = "jabuka" if pravi == 0 else "banana"
    if predvidjanje == pravi_naziv:
        tacno += 1
    print(f"  Primer {i+1}: AI kaže -> {predvidjanje}, tačno je -> {pravi_naziv}")

procenat = (tacno / len(primeri)) * 100
print(f"Ukupna tačnost AI modela: {tacno}/{len(primeri)} = {procenat:.0f}%\n")
# =====================================================================
# TESTIRANJE NA NOVOM VOĆU I KORISNIČKI UNOS
# =====================================================================

print("--- Testiranje na novom voću (koje AI nikad nije video) ---")

# Crvena, mala voćka
novo1 = [3, 1]
rez1, _ = ai_pogodi(novo1, primeri, odgovori)
print(f"1. Test (Crvena, mala): AI kaže da je to --> {rez1}")

# Žuta, velika voćka
novo2 = [1, 3]
rez2, _ = ai_pogodi(novo2, primeri, odgovori)
print(f"2. Test (Žuta, velika): AI kaže da je to --> {rez2}\n")

print("--- Sada ti biraš! Unesi osobine za AI ---")
moja_boja = int(input("Unesi boju (1=žuta, 2=zelena, 3=crvena): "))
moja_vel = int(input("Unesi veličinu (1=mala, 2=srednja, 3=velika): "))

moje_voce = [moja_boja, moja_vel]
rez_korisnik, _ = ai_pogodi(moje_voce, primeri, odgovori)

print(f"\n[AI REZULTAT]: AI misli da je tvoje voće: {rez_korisnik.upper()}!")
print("Da li si saglasan sa procenom veštačke inteligencije?")
# =====================================================================
# DEO 3: PROŠIRENI AI (Dodajemo novu vrstu - Narandžu)
# =====================================================================

# Proširujemo bazu podataka novim primerima za narandžu (boja 4 = narandžasta)
primeri_prosireni = primeri + [
    [4, 2],  # narandžasta, srednja  --> narandža
    [4, 1],  # narandžasta, mala     --> narandža
    [4, 3],  # narandžasta, velika   --> narandža
]

# Rečnik odgovora proširujemo brojem 2 (2 označava narandžu)
odgovori_prosireni = odgovori + [2, 2, 2]


def ai_pogodi_v2(novo_voce, primeri, odgovori):
    najmanje_r = 999
    najblizi_odgovor = None

    for p, o in zip(primeri, odgovori):
        r = rastojanje(novo_voce, p)  # Koristimo istu funkciju za rastojanje
        if r < najmanje_r:
            najmanje_r = r
            najblizi_odgovor = o

    # Rečnik za mapiranje brojeva u tekst
    nazivi = {0: "jabuka", 1: "banana", 2: "narandža"}
    return nazivi[najblizi_odgovor]


# Brzi test proširenog modela
test_narandza = ai_pogodi_v2([4, 2], primeri_prosireni, odgovori_prosireni)
test_jabuka = ai_pogodi_v2([3, 2], primeri_prosireni, odgovori_prosireni)

print("--- Testiranje proširenog AI modela (v2) ---")
print(f"Unos [4, 2] (narandžasta, srednja) --> AI kaže: {test_narandza}")
print(f"Unos [3, 2] (crvena, srednja)       --> AI kaže: {test_jabuka}")
```
